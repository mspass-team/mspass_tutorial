# Exporting Metadata Collections
This notebook uses pymongo to export database collections common to the entire data set I'm assembling.   The json files it produces can be downloaded and the inverse performed on the local system.  (a different notebook)

```python
import json
import pymongo
from bson import json_util

from mspasspy.client import Client
mspass_client=Client()
db = mspass_client.get_database("ANF48")
```

```python
# list of collections to export
export_list = ["site","channel","source","arrival_css30"]
for collection in export_list:
    cursor = db[collection].find({})
    output_file = f"{db.name}_{collection}.json"
    with open(output_file,"w") as fh:
        # Stream a JSON array without materializing the full collection.
        fh.write("[\n")
        for index, document in enumerate(cursor):
            if index:
                fh.write(",\n")
            fh.write(json_util.dumps(document))
        fh.write("\n]\n")
    print("Wrote data for collection {} to file={}".format(collection,output_file))
```
